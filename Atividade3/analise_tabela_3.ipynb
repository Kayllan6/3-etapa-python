# Análise da Tabela 3 - Graduação por Região, Sexo e Área de Formação

```python

import pandas as pd


caminho_arquivo = 'Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx'
df_raw = pd.read_excel(caminho_arquivo)

regioes = ['Norte', 'Nordeste', 'Sudeste', 'Sul', 'Centro-Oeste']
df_regioes = df_raw[df_raw.iloc[:, 0].isin(regioes)].copy()


tabela_1 = df_regioes.iloc[:, [0, 1, 2, 3]].copy()
tabela_1.columns = ['Região', 'Total', 'Homens', 'Mulheres']
for col in ['Total', 'Homens', 'Mulheres']:
    tabela_1[col] = pd.to_numeric(tabela_1[col])


tabela_2 = df_regioes.iloc[:, [0, 4, 5, 6]].copy()
tabela_2.columns = ['Região', 'Total', 'Homens', 'Mulheres']
for col in ['Total', 'Homens', 'Mulheres']:
    tabela_2[col] = pd.to_numeric(tabela_2[col])


tabela_3 = df_regioes.iloc[:, [0, 7, 8, 9]].copy()
tabela_3.columns = ['Região', 'Total', 'Homens', 'Mulheres']
for col in ['Total', 'Homens', 'Mulheres']:
    tabela_3[col] = pd.to_numeric(tabela_3[col])


tabela_1.to_excel('tabela_1_curso_superior.xlsx', index=False)
tabela_2.to_excel('tabela_2_stem.xlsx', index=False)
tabela_3.to_excel('tabela_3_educacao_saude.xlsx', index=False)

with pd.ExcelWriter('Tabelas_Divisao_Regioes.xlsx') as writer:
    tabela_1.to_excel(writer, sheet_name='Curso_Superior', index=False)
    tabela_2.to_excel(writer, sheet_name='STEM', index=False)
    tabela_3.to_excel(writer, sheet_name='Educacao_Saude', index=False)

print("Tabelas salvas com sucesso!")
```
