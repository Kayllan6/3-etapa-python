# Aula — Tabelas 5 + Tabela 1.1.1




```python
import pandas as pd


arquivo_2012 = "Tabela5-sem_emprego_2012.csv"
arquivo_2026 = "Tabela5-sem_emprego_2026.csv"
arquivo_horas = "Tabela 1.1.1.xls"


t2012 = pd.read_csv(arquivo_2012, sep=";", encoding="utf-8-sig")
t2026 = pd.read_csv(arquivo_2026, sep=";", encoding="utf-8-sig")


display(t2012.head())
display(t2026.head())
```

```python

comp = t2012.merge(
    t2026,
    on=["Sigla", "Código", "Estado"],
    how="inner"
)

comp.head()
```

```python

tabela_horas = pd.read_excel(
    arquivo_horas,
    sheet_name="2022",
    header=None
)

tabela_horas.head(10)
```

```python


horas = tabela_horas.iloc[8:, [0, 1]].copy()
horas.columns = ["Estado", "Horas_semanais_total"]


estados = comp["Estado"].tolist()
horas = horas[horas["Estado"].isin(estados)].copy()


horas["Horas_semanais_total"] = pd.to_numeric(
    horas["Horas_semanais_total"],
    errors="coerce"
)

horas.head()
```

```python

cruzado = comp.merge(
    horas,
    on="Estado",
    how="inner"
)

cruzado.head()
```

```python

colunas_numericas = [
    "Desocupados - homens (2012 T1)",
    "Desocupados - mulheres (2012 T1)",
    "Desocupados - homens (2026 T1)",
    "Desocupados - mulheres (2026 T1)"
]

for coluna in colunas_numericas:
    cruzado[coluna] = (
        cruzado[coluna]
        .astype(str)
        .str.replace(",", ".", regex=False)
        .astype(float)
    )

cruzado["Diferença homens 2026-2012"] = (
    cruzado["Desocupados - homens (2026 T1)"]
    - cruzado["Desocupados - homens (2012 T1)"]
)

cruzado["Diferença mulheres 2026-2012"] = (
    cruzado["Desocupados - mulheres (2026 T1)"]
    - cruzado["Desocupados - mulheres (2012 T1)"]
)


cruzado["Diferença mulheres-homens 2012"] = (
    cruzado["Desocupados - mulheres (2012 T1)"]
    - cruzado["Desocupados - homens (2012 T1)"]
)

cruzado["Diferença mulheres-homens 2026"] = (
    cruzado["Desocupados - mulheres (2026 T1)"]
    - cruzado["Desocupados - homens (2026 T1)"]
)

cruzado

```

```python

cruzado = cruzado[
    [
        "Sigla",
        "Código",
        "Estado",
        "Desocupados - homens (2012 T1)",
        "Desocupados - mulheres (2012 T1)",
        "Desocupados - homens (2026 T1)",
        "Desocupados - mulheres (2026 T1)",
        "Diferença homens 2026-2012",
        "Diferença mulheres 2026-2012",
        "Diferença mulheres-homens 2012",
        "Diferença mulheres-homens 2026",
        "Horas_semanais_total"
    ]
]

cruzado
```

```python


cruzado.to_csv(
    "geral.csv",
    sep=";",
    index=False,
    encoding="utf-8-sig"
)

cruzado.to_excel(
    "geral.xlsx",
    index=False
)

print("Arquivos salvos: geral.csv e geral.xlsx")
```
